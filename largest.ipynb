```python
def find_largest_number(numbers_list: list[float]) ->float:
    if not isinstance(numbers_list,list):
        raise TypeError("Input must be a list of num:")
    if not numbers_list:
        raise ValueError("List can not be empty")
    for num in numbers_list:
        if not isinstance(num,(int,float)):
            raise TypeError("All elements must be numeric")
    largest = numbers_list[0]
    for num in numbers_list:
        if num>largest:
            largest=num
    return largest
if __name__ == "__main__":
    try:
        user_input = input("Enter numbers in the list:")
        numbers_list = [int(item) for item in user_input.split()]
        result = find_largest_number(numbers_list)
        print("Largest numbers in the list is:",result)
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
        
    
```

```python
def is_prime_number(number: int)->bool:
    if not isinstance(number, int):
        raise TypeError("Input must be an integer.")
    if number < 2:
        raise ValueError("Prime number are greater then or equal to 2.")
    for i in range(2, int(number**0.5) +1):
        if number % i == 0:
            return False
    else:
        return True
if __name__ == "__main__":
    try:
        prime_number=[]
        lower_limit=int(input("Enter lower limit"))
        upper_limit=int(input("Enter upper limit"))
        for i in range(lower_limit,upper_limit):
            if is_prime_number(i):
                print(f"{i} prime number")
                prime_number.append(i)
         print(prime_number)

    except Exception as e:
        print(e)
            
            
```

```python
def factorial_finder(num: int) -> int:
    if isinstance(num,bool):
        raise TypeError("Boolean values are not allowed")
    if not isinstance(num,int):
        raise TypeError("input must be an integer")

    if num < 0:
        raise ValueError("Factorial is not define as negative")

    factorial = 1
    for i in range(1,num+1):
        factorial = factorial * i
    return(factorial)

if __name__ == " __main__":
    try:
        number = input("Enter your number to find factrial:")
        number = int(number)
        print("factorial:",factorial_finder(number))

    except ValueError as e:
        print(e)
    except TypeError as e:
        print("Error",e)
    
        
    




    
        
    


    

    
    
   
    
```
